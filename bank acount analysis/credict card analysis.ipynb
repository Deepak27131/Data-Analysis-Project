# Examine the distribution of unique categories in categorical columns

```python
categorical_cols = df[['BRId','GenderId','IAId','Amount of Credit Cards', 'Nationality','Occupation','Fee Structure','Loyalty Classification', 'Properties Owned','Risk Weighting','Income Band']].columns           

for col in categorical_cols:
    print(f"value Counts for '{col}':")
    display(df[col].value_counts())
```

## Univariate Analysis

```python

for i, predictor in enumerate(df[['BRId','GenderId','IAId','Amount of Credit Cards', 'Nationality','Occupation','Fee Structure','Loyalty Classification', 'Properties Owned','Risk Weighting','Income Band']].columns ):
    plt.figure(i)
    sns.countplot(data=df, x = predictor)
```

## Baivariate Analysis

```python
for i, predictor in enumerate(df[['BRId','GenderId','IAId','Amount of Credit Cards', 'Nationality','Occupation','Fee Structure','Loyalty Classification', 'Properties Owned','Risk Weighting','Income Band']].columns ):
    plt.figure(i)
    sns.countplot(data=df, x = predictor , hue = 'GenderId')
```

```python
for i, predictor in enumerate(df[['BRId','GenderId','IAId','Amount of Credit Cards', 'Nationality','Occupation','Fee Structure','Loyalty Classification', 'Properties Owned','Risk Weighting','Icome Band']].columns ):
    plt.figure(i)
    sns.countplot(data=df, x = predictor , hue = 'Nationality')
```

```python
# his plot of value countr  of different Occupation

for col in categorical_cols:
    if col == 'Occupation':
        continue
    plt.figure(figsize=(8,4))
    sns.histplot(df[col])
    plt.title('Histogram of Occupation Count')
    plt.xlabel(col)
    plt.ylabel('count')
    plt.show()
```

## NumericaL ANALYSIS 


```python
numerical_cols = ['Estimated Income', 'Superannuation Savings','Credit Card Balance','Bank Deposits','Checking Accounts','Saving Accounts','Foreign Currency Account','Business Lending']

#univariate analysis and visulization
plt.figure(figsize = (20,15))
for i,col in enumerate(numerical_cols):
    plt.subplot(4,3,i+1)
    sns.histplot(df[col],kde=True)
plt.show()
```

## Heat Map

```python
numerical_cols = ['Estimated Income', 'Superannuation Savings','Credit Card Balance','Bank Deposits','Checking Accounts','Saving Accounts','Foreign Currency Account','Business Lending']

correlation_matrix = df[numerical_cols].corr()
plt.figure(figsize = (12,12))
sns.heatmap(correlation_matrix, annot=True, cmap = 'crest',fmt=".2f")
plt.title("Correlation Matrix ")
plt.show()
```

### Insight of EDA 

1. The strongest positive correlation occur among "Bank Deposits" with "Checking Account ","Saving Account" and "Foreign Currency Account" ,indicating that  customer  who
maintain high balances in one account often hold substantial account/founds across other account as well.
